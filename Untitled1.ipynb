Data Imputation

```python

num_cols = [
    'Age',
    'RoomService',
    'FoodCourt',
    'ShoppingMall',
    'Spa',
    'VRDeck'
]

for col in num_cols:
    train[col] = train[col].fillna(train[col].median())
    test[col] = test[col].fillna(test[col].median())
```

```python
cat_cols = [
    'HomePlanet',
    'CryoSleep',
    'Cabin',
    'Destination',
    'VIP'
]

for col in cat_cols:
    train[col] = train[col].fillna(train[col].mode()[0])
    test[col] = test[col].fillna(test[col].mode()[0])
```

```python
train[['Deck','CabinNum','Side']] = train['Cabin'].str.split('/', expand=True)

test[['Deck','CabinNum','Side']] = test['Cabin'].str.split('/', expand=True)
```

```python
train.drop('Cabin', axis=1, inplace=True)
test.drop('Cabin', axis=1, inplace=True)
```

```python
train.drop('Name', axis=1, inplace=True)
test.drop('Name', axis=1, inplace=True)
```

```python
test_ids = test['PassengerId']
```

```python
train.drop('PassengerId', axis=1, inplace=True)
test.drop('PassengerId', axis=1, inplace=True)
```

```python
all_data = pd.concat([train.drop('Transported', axis=1), test])

all_data = pd.get_dummies(all_data)

X = all_data.iloc[:len(train)]
X_test = all_data.iloc[len(train):]
```

```python
y = train['Transported']
```

```python


model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X, y)

print("Training Complete!")
```
